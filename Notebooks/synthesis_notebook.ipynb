            <strong style="color: #00d9ff;">Problem:</strong> SVMs output a "distance to hyperplane," not a probability.<br>
            <strong style="color: #00d9ff;">Solution:</strong> We apply <strong>Sigmoid Calibration</strong> to map these distances to a confident 0-100% risk score.

```python
# Step 5: Probability Calibration (The "Fix")
from sklearn.calibration import CalibratedClassifierCV
import numpy as np

print("\n" + "=" * 70)
print("PROBABILITY CALIBRATION (SIGMOID SCALING)")
print("=" * 70)

# 1. Safety Check: Ensure the base model is fitted
try:
    model.predict(X_test_scaled[:1])
    print("✅ Base SGD Model is already fitted.")
except:
    print("⚠️ Base model not fitted. Retraining now...")
    model.fit(X_train_scaled, y_train)

# 2. Calibration
# We use cv='prefit' because we want to calibrate the ALREADY trained model
# method='sigmoid' (Platt Scaling) is best for SVMs
calibrator = CalibratedClassifierCV(model, method='sigmoid', cv='prefit')

# Fit the calibrator
# Note: With cv='prefit', this step actually just learns the mapping from 
# the base model's decision function to probabilities. It doesn't retrain the base model.
calibrator.fit(X_train_scaled, y_train)

# 3. Predict Probabilities
# We take [:, 1] for the probability of Class 1 (Malignant)
y_probs = calibrator.predict_proba(X_test_scaled)[:, 1]

print(f"✅ Probabilities Generated.")
print(f"   Sample scores: {np.round(y_probs[:5], 2)}")
```

```python
# Step 6: Risk Stratification (Resource Allocation)
import pandas as pd

# Create the Patient List
patient_ids = [f'P-{1000+i}' for i in range(len(y_probs))]
allocation_df = pd.DataFrame({'Patient_ID': patient_ids, 'Probability': y_probs})

# Stratification Logic: Use Quantiles to match Hospital Capacity
# Low Risk: Bottom 50% of cases
# Medium Risk: Next 30%
# High Risk: Top 20% (The most urgent cases)
allocation_df['Risk_Group'] = pd.qcut(
    allocation_df['Probability'], 
    q=[0, 0.5, 0.8, 1.0], 
    labels=['🟢 Low', '🟡 Medium', '🔴 High']
)

# Business Output: The Priority List
high_risk_cases = allocation_df[allocation_df['Risk_Group'] == '🔴 High'].sort_values('Probability', ascending=False)

print("🏥 PRIORITY QUEUE (High Risk Cases Only)")
print(high_risk_cases.head(10).to_string(index=False))
```
