```python
print("\n" + "="*80)
print("SECTION 2: MISSING VALUES ANALYSIS")
print("="*80 + "\n")

# Calculate missing values
metabric_missing_cols = metabric_df.isnull().sum()
metabric_missing_cols = metabric_missing_cols[metabric_missing_cols > 0].sort_values(ascending=False)

print(f"Columns with missing values: {len(metabric_missing_cols)}")
print(f"\nTop 15 columns with missing values:")
for i, (col, count) in enumerate(metabric_missing_cols.head(15).items(), 1):
    metabric_pct = (count / len(metabric_df)) * 100
    print(f"{i:2d}. {col:40s}: {count:5d} ({metabric_pct:5.1f}%)")

# Visualize missing values
metabric_fig, metabric_axes = plt.subplots(1, 2, figsize=(16, 6))
metabric_fig.patch.set_facecolor('#0a0a0a')
metabric_fig.suptitle('Missing Values Analysis', fontsize=14, fontweight='bold', color='#ffffff')

metabric_top_missing = metabric_missing_cols.head(20)
metabric_axes[0].barh(range(len(metabric_top_missing)), metabric_top_missing.values, 
                      color='#ff1493', edgecolor='#ffffff', linewidth=0.5, alpha=0.8)
metabric_axes[0].set_yticks(range(len(metabric_top_missing)))
metabric_axes[0].set_yticklabels(metabric_top_missing.index, fontsize=9, color='#bbbbbb')
metabric_axes[0].set_xlabel('Number of Missing Values', color='#ffffff')
metabric_axes[0].set_title('Top 20 Columns with Missing Values', color='#ffffff', pad=10)
metabric_axes[0].axvline(x=len(metabric_df)*0.8, color='#00d9ff', linestyle='--', linewidth=2, label='80% threshold')
metabric_axes[0].legend(facecolor='#141414', edgecolor='#333333')
metabric_axes[0].grid(alpha=0.2, axis='x')
metabric_axes[0].set_facecolor('#141414')

metabric_missing_pct = (metabric_df.isnull().sum() / len(metabric_df) * 100).sort_values(ascending=False)
metabric_missing_pct = metabric_missing_pct[metabric_missing_pct > 0]
metabric_axes[1].hist(metabric_missing_pct, bins=30, color='#00d9ff', edgecolor='#ffffff', 
                      linewidth=0.5, alpha=0.8)
metabric_axes[1].axvline(x=80, color='#ff1493', linestyle='--', linewidth=2, label='80% threshold')
metabric_axes[1].set_xlabel('Missing Percentage (%)', color='#ffffff')
metabric_axes[1].set_ylabel('Number of Columns', color='#ffffff')
metabric_axes[1].set_title('Distribution of Missing Percentages', color='#ffffff', pad=10)
metabric_axes[1].legend(facecolor='#141414', edgecolor='#333333')
metabric_axes[1].grid(alpha=0.2)
metabric_axes[1].set_facecolor('#141414')

plt.tight_layout()
plt.show()

```

```python
print("\n" + "="*80)
print("SECTION 4: NUMERICAL VARIABLES DISTRIBUTION")
print("="*80 + "\n")

metabric_fig2, metabric_axes2 = plt.subplots(2, 3, figsize=(18, 10))
metabric_fig2.patch.set_facecolor('#0a0a0a')
metabric_fig2.suptitle('Numerical Variables Distribution', fontsize=16, fontweight='bold', color='#ffffff')

if 'age_at_diagnosis' in metabric_df.columns:
    metabric_age_data = metabric_df['age_at_diagnosis'].dropna()
    metabric_axes2[0, 0].hist(metabric_age_data, bins=30, color='#00d9ff', edgecolor='#ffffff', 
                              linewidth=0.5, alpha=0.8)
    metabric_axes2[0, 0].axvline(metabric_age_data.mean(), color='#ff1493', linestyle='--', 
                                  linewidth=2, label=f'Mean: {metabric_age_data.mean():.1f}')
    metabric_axes2[0, 0].set_xlabel('Age at diagnosis', color='#ffffff')
    metabric_axes2[0, 0].set_ylabel('Frequency', color='#ffffff')
    metabric_axes2[0, 0].set_title(f'Age Distribution\n(Missing: {metabric_df["age_at_diagnosis"].isnull().sum()})', 
                                   color='#ffffff', pad=10)
    metabric_axes2[0, 0].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes2[0, 0].grid(alpha=0.2)
    metabric_axes2[0, 0].set_facecolor('#141414')

if 'tumor_size' in metabric_df.columns:
    metabric_size_data = pd.to_numeric(metabric_df['tumor_size'], errors='coerce').dropna()
    metabric_axes2[0, 1].hist(metabric_size_data, bins=30, color='#ff1493', edgecolor='#ffffff', 
                              linewidth=0.5, alpha=0.8)
    metabric_axes2[0, 1].axvline(metabric_size_data.median(), color='#00d9ff', linestyle='--', 
                                  linewidth=2, label=f'Median: {metabric_size_data.median():.1f}mm')
    metabric_axes2[0, 1].set_xlabel('Tumor size (mm)', color='#ffffff')
    metabric_axes2[0, 1].set_ylabel('Frequency', color='#ffffff')
    metabric_axes2[0, 1].set_title(f'Tumor Size Distribution\n(Missing: {metabric_df["tumor_size"].isnull().sum()})', 
                                   color='#ffffff', pad=10)
    metabric_axes2[0, 1].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes2[0, 1].grid(alpha=0.2)
    metabric_axes2[0, 1].set_facecolor('#141414')

if 'lymph_nodes_examined_positive' in metabric_df.columns:
    metabric_ln_data = pd.to_numeric(metabric_df['lymph_nodes_examined_positive'], errors='coerce').dropna()
    metabric_axes2[0, 2].hist(metabric_ln_data, bins=30, color='#9d4edd', edgecolor='#ffffff', 
                              linewidth=0.5, alpha=0.8)
    metabric_axes2[0, 2].axvline(metabric_ln_data.median(), color='#00d9ff', linestyle='--', 
                                  linewidth=2, label=f'Median: {metabric_ln_data.median():.1f}')
    metabric_axes2[0, 2].set_xlabel('Positive lymph nodes', color='#ffffff')
    metabric_axes2[0, 2].set_ylabel('Frequency', color='#ffffff')
    metabric_axes2[0, 2].set_title(f'Lymph Nodes Distribution\n(Missing: {metabric_df["lymph_nodes_examined_positive"].isnull().sum()})', 
                                   color='#ffffff', pad=10)
    metabric_axes2[0, 2].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes2[0, 2].grid(alpha=0.2)
    metabric_axes2[0, 2].set_facecolor('#141414')

if 'overall_survival_months' in metabric_df.columns:
    metabric_os_data = pd.to_numeric(metabric_df['overall_survival_months'], errors='coerce').dropna()
    metabric_axes2[1, 0].hist(metabric_os_data, bins=30, color='#f72585', edgecolor='#ffffff', 
                              linewidth=0.5, alpha=0.8)
    metabric_axes2[1, 0].axvline(metabric_os_data.median(), color='#00d9ff', linestyle='--', 
                                  linewidth=2, label=f'Median: {metabric_os_data.median():.1f} months')
    metabric_axes2[1, 0].set_xlabel('Overall Survival (months)', color='#ffffff')
    metabric_axes2[1, 0].set_ylabel('Frequency', color='#ffffff')
    metabric_axes2[1, 0].set_title(f'Survival Distribution\n(Missing: {metabric_df["overall_survival_months"].isnull().sum()})', 
                                   color='#ffffff', pad=10)
    metabric_axes2[1, 0].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes2[1, 0].grid(alpha=0.2)
    metabric_axes2[1, 0].set_facecolor('#141414')

if 'nottingham_prognostic_index' in metabric_df.columns:
    metabric_npi_data = pd.to_numeric(metabric_df['nottingham_prognostic_index'], errors='coerce').dropna()
    metabric_axes2[1, 1].hist(metabric_npi_data, bins=30, color='#ffa500', edgecolor='#ffffff', 
                              linewidth=0.5, alpha=0.8)
    metabric_axes2[1, 1].axvline(metabric_npi_data.mean(), color='#00d9ff', linestyle='--', 
                                  linewidth=2, label=f'Mean: {metabric_npi_data.mean():.2f}')
    metabric_axes2[1, 1].set_xlabel('Nottingham Prognostic Index', color='#ffffff')
    metabric_axes2[1, 1].set_ylabel('Frequency', color='#ffffff')
    metabric_axes2[1, 1].set_title(f'NPI Distribution\n(Missing: {metabric_df["nottingham_prognostic_index"].isnull().sum()})', 
                                   color='#ffffff', pad=10)
    metabric_axes2[1, 1].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes2[1, 1].grid(alpha=0.2)
    metabric_axes2[1, 1].set_facecolor('#141414')

# Hide unused subplot
metabric_axes2[1, 2].axis('off')
metabric_axes2[1, 2].set_facecolor('#0a0a0a')

plt.tight_layout()
plt.show()
```

```python
print("\n" + "="*80)
print("SECTION 2: DATA IMPUTATION")
print("="*80 + "\n")

metabric_clean = metabric_selected.copy()
print(f"Missing values before: {metabric_clean.isnull().sum().sum()}")

# Numeric imputation (median)
metabric_numeric_cols = ['age_at_diagnosis', 'tumor_size', 'lymph_nodes_examined_positive',
                         'nottingham_prognostic_index', 'overall_survival_months', 'mutation_count']
metabric_available_numeric = [col for col in metabric_numeric_cols if col in metabric_clean.columns]

for col in metabric_available_numeric:
    if metabric_clean[col].isnull().sum() > 0:
        metabric_clean[col] = pd.to_numeric(metabric_clean[col], errors='coerce')
        metabric_clean[col].fillna(metabric_clean[col].median(), inplace=True)

# Categorical imputation (mode)
metabric_categorical_cols = ['tumor_stage', 'neoplasm_histologic_grade', 'cellularity',
                              'er_status', 'her2_status', 'pr_status',
                              'pam50_+_claudin-low_subtype', 'integrative_cluster',
                              '3-gene_classifier_subtype']
metabric_available_categorical = [col for col in metabric_categorical_cols if col in metabric_clean.columns]

for col in metabric_available_categorical:
    if metabric_clean[col].isnull().sum() > 0:
        metabric_mode = metabric_clean[col].mode()[0] if len(metabric_clean[col].mode()) > 0 else 'Unknown'
        metabric_clean[col].fillna(metabric_mode, inplace=True)

# Mutation imputation (0)
metabric_mutation_cols = [col for col in metabric_clean.columns if col.endswith('_mut')]
for col in metabric_mutation_cols:
    metabric_clean[col].fillna(0, inplace=True)

print(f"Missing values after: {metabric_clean.isnull().sum().sum()}")
```

         Outlier Detection & Treatment
    </h3>
    <div style="background: #141414; padding: 10px 15px; border-radius: 4px; border-left: 1px solid #333333;">
        <p style="font-size: 13px; color: #bbbbbb; line-height: 1.5; margin: 0; font-family: sans-serif;">
            <strong style="color: #00d9ff;">Objective:</strong> Identify and cap outliers using IQR method, treating only variables with >5% outliers to preserve data integrity.
        </p>
    </div>
</div>

```python
print("\n" + "="*80)
print("SECTION 3: OUTLIER DETECTION & TREATMENT")
print("="*80 + "\n")

def detect_and_cap_outliers(df, column):
    data = df[column].copy()
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((data < lower) | (data > upper)).sum()
    pct = (outliers / len(data)) * 100
    
    data_capped = data.clip(lower=lower, upper=upper)
    return data_capped, {'n_outliers': outliers, 'pct': pct}

metabric_outlier_cols = ['age_at_diagnosis', 'tumor_size', 
                         'lymph_nodes_examined_positive', 'overall_survival_months']
metabric_available_outlier = [col for col in metabric_outlier_cols if col in metabric_clean.columns]

# Visualize outliers before treatment
metabric_fig1, metabric_axes1 = plt.subplots(2, 2, figsize=(16, 10))
metabric_fig1.patch.set_facecolor('#0a0a0a')
metabric_fig1.suptitle('Outlier Detection & Treatment', fontsize=16, fontweight='bold', color='#ffffff')

for idx, col in enumerate(metabric_available_outlier[:4]):
    row, col_idx = idx // 2, idx % 2
    ax = metabric_axes1[row, col_idx]
    
    data_before = metabric_clean[col]
    data_after, stats = detect_and_cap_outliers(metabric_clean, col)
    
    # Only treat if >5% outliers
    if stats['pct'] > 5:
        metabric_clean[col] = data_after
        status = "TREATED"
        color = '#ff1493'
    else:
        status = "OK"
        color = '#00d9ff'
    
    # Box plot comparison
    ax.boxplot([data_before, data_after], labels=['Before', 'After'],
               patch_artist=True,
               boxprops=dict(facecolor=color, alpha=0.6, edgecolor='#ffffff'),
               medianprops=dict(color='#ffffff', linewidth=2),
               whiskerprops=dict(color='#bbbbbb'),
               capprops=dict(color='#bbbbbb'))
    
    ax.set_title(f'{col}\n{status}: {stats["n_outliers"]} outliers ({stats["pct"]:.1f}%)',
                color='#ffffff', pad=10)
    ax.set_ylabel('Value', color='#ffffff')
    ax.set_facecolor('#141414')
    ax.grid(alpha=0.2, axis='y')

plt.tight_layout()
plt.show()
```

```python
print("\n" + "="*80)
print("SECTION 4: CATEGORICAL ENCODING")
print("="*80 + "\n")

# Binary encoding
metabric_binary_cols = {
    'er_status': {'Positive': 1, 'Positve': 1, 'Negative': 0},
    'her2_status': {'Positive': 1, 'Negative': 0},
    'pr_status': {'Positive': 1, 'Negative': 0}
}

for col, mapping in metabric_binary_cols.items():
    if col in metabric_clean.columns:
        metabric_clean[col + '_binary'] = metabric_clean[col].map(mapping)
        metabric_clean[col + '_binary'].fillna(metabric_clean[col + '_binary'].mode()[0], inplace=True)

# Ordinal encoding
if 'tumor_stage' in metabric_clean.columns:
    stage_map = {'0': 0, '1': 1, '2': 2, '3': 3, '4': 4}
    metabric_clean['tumor_stage_encoded'] = metabric_clean['tumor_stage'].astype(str).str.replace('.0', '').map(stage_map)
    metabric_clean['tumor_stage_encoded'].fillna(metabric_clean['tumor_stage_encoded'].median(), inplace=True)

if 'neoplasm_histologic_grade' in metabric_clean.columns:
    grade_map = {'1': 1, '2': 2, '3': 3}
    metabric_clean['neoplasm_histologic_grade_encoded'] = metabric_clean['neoplasm_histologic_grade'].astype(str).str.replace('.0', '').map(grade_map)
    metabric_clean['neoplasm_histologic_grade_encoded'].fillna(metabric_clean['neoplasm_histologic_grade_encoded'].mode()[0], inplace=True)

if 'cellularity' in metabric_clean.columns:
    cell_map = {'Low': 0, 'Moderate': 1, 'High': 2}
    metabric_clean['cellularity_encoded'] = metabric_clean['cellularity'].map(cell_map)
    metabric_clean['cellularity_encoded'].fillna(metabric_clean['cellularity_encoded'].mode()[0], inplace=True)

# Label encoding for nominal variables
from sklearn.preprocessing import LabelEncoder
metabric_nominal_cols = ['pam50_+_claudin-low_subtype', 'integrative_cluster', '3-gene_classifier_subtype']

for col in metabric_nominal_cols:
    if col in metabric_clean.columns:
        metabric_clean[col].fillna('Unknown', inplace=True)
        le = LabelEncoder()
        metabric_clean[col + '_encoded'] = le.fit_transform(metabric_clean[col].astype(str))

# Survival encoding
if 'overall_survival' in metabric_clean.columns:
    survival_map = {'LIVING': 1, 'Living': 1, 'DECEASED': 0, 'Deceased': 0}
    metabric_clean['overall_survival_binary'] = metabric_clean['overall_survival'].map(survival_map).fillna(0)

if 'death_from_cancer' in metabric_clean.columns:
    death_map = {'Living': 0, 'Died of Disease': 1, 'Died of Other Causes': 0}
    metabric_clean['death_from_cancer_binary'] = metabric_clean['death_from_cancer'].map(death_map).fillna(0)

print("✓ Encoding completed")
```

```python
print("\n" + "="*80)
print("SECTION 5: FEATURE ENGINEERING")
print("="*80 + "\n")

# Hormone Receptor Score
if all(col in metabric_clean.columns for col in ['er_status_binary', 'pr_status_binary']):
    metabric_clean['hormone_receptor_score'] = metabric_clean['er_status_binary'] + metabric_clean['pr_status_binary']

# Triple Negative
if all(col in metabric_clean.columns for col in ['er_status_binary', 'her2_status_binary', 'pr_status_binary']):
    metabric_clean['triple_negative'] = (
        (metabric_clean['er_status_binary'] == 0) & 
        (metabric_clean['her2_status_binary'] == 0) & 
        (metabric_clean['pr_status_binary'] == 0)
    ).astype(int)

# Size Category
if 'tumor_size' in metabric_clean.columns:
    metabric_clean['size_category'] = (metabric_clean['tumor_size'] > metabric_clean['tumor_size'].median()).astype(int)

# Grade-Stage Interaction
if all(col in metabric_clean.columns for col in ['neoplasm_histologic_grade_encoded', 'tumor_stage_encoded']):
    metabric_clean['grade_stage_interaction'] = (
        metabric_clean['neoplasm_histologic_grade_encoded'] * metabric_clean['tumor_stage_encoded']
    )

# High Risk Score
if all(col in metabric_clean.columns for col in ['neoplasm_histologic_grade_encoded', 'tumor_stage_encoded']):
    metabric_clean['high_risk'] = (
        metabric_clean['neoplasm_histologic_grade_encoded'] + 
        metabric_clean['tumor_stage_encoded']
    )

# Visualize engineered features
metabric_fig2, metabric_axes2 = plt.subplots(2, 3, figsize=(18, 10))
metabric_fig2.patch.set_facecolor('#0a0a0a')
metabric_fig2.suptitle('Engineered Features', fontsize=16, fontweight='bold', color='#ffffff')

if 'hormone_receptor_score' in metabric_clean.columns:
    hr_counts = metabric_clean['hormone_receptor_score'].value_counts().sort_index()
    metabric_axes2[0, 0].bar(hr_counts.index.astype(str), hr_counts.values, 
                             color='#ff1493', edgecolor='#ffffff', linewidth=0.5, alpha=0.8)
    metabric_axes2[0, 0].set_xlabel('Score', color='#ffffff')
    metabric_axes2[0, 0].set_ylabel('Count', color='#ffffff')
    metabric_axes2[0, 0].set_title('Hormone Receptor Score', color='#ffffff', pad=10)
    metabric_axes2[0, 0].set_facecolor('#141414')
    metabric_axes2[0, 0].grid(alpha=0.2, axis='y')

if 'triple_negative' in metabric_clean.columns:
    tn_counts = metabric_clean['triple_negative'].value_counts()
    metabric_axes2[0, 1].pie(tn_counts.values, labels=['Not TN', 'Triple Negative'], 
                             autopct='%1.1f%%', colors=['#00d9ff', '#ff1493'])
    metabric_axes2[0, 1].set_title('Triple Negative Distribution', color='#ffffff', pad=10)

if 'size_category' in metabric_clean.columns:
    size_cat_counts = metabric_clean['size_category'].value_counts()
    metabric_axes2[0, 2].bar(['Small (0)', 'Large (1)'], 
                             [size_cat_counts.get(0, 0), size_cat_counts.get(1, 0)],
                             color='#9d4edd', edgecolor='#ffffff', linewidth=0.5, alpha=0.8)
    metabric_axes2[0, 2].set_ylabel('Count', color='#ffffff')
    metabric_axes2[0, 2].set_title('Tumor Size Category', color='#ffffff', pad=10)
    metabric_axes2[0, 2].set_facecolor('#141414')
    metabric_axes2[0, 2].grid(alpha=0.2, axis='y')

if 'grade_stage_interaction' in metabric_clean.columns:
    gs_counts = metabric_clean['grade_stage_interaction'].value_counts().sort_index()
    metabric_axes2[1, 0].bar(gs_counts.index.astype(str), gs_counts.values, 
                             color='#00d9ff', edgecolor='#ffffff', linewidth=0.5, alpha=0.8)
    metabric_axes2[1, 0].set_xlabel('Interaction Score', color='#ffffff')
    metabric_axes2[1, 0].set_ylabel('Count', color='#ffffff')
    metabric_axes2[1, 0].set_title('Grade × Stage Interaction', color='#ffffff', pad=10)
    metabric_axes2[1, 0].set_facecolor('#141414')
    metabric_axes2[1, 0].grid(alpha=0.2, axis='y')

if 'high_risk' in metabric_clean.columns:
    risk_counts = metabric_clean['high_risk'].value_counts().sort_index()
    metabric_axes2[1, 1].bar(risk_counts.index.astype(str), risk_counts.values, 
                             color='#ffa500', edgecolor='#ffffff', linewidth=0.5, alpha=0.8)
    metabric_axes2[1, 1].set_xlabel('Risk Level', color='#ffffff')
    metabric_axes2[1, 1].set_ylabel('Count', color='#ffffff')
    metabric_axes2[1, 1].set_title('Risk Score Distribution', color='#ffffff', pad=10)
    metabric_axes2[1, 1].set_facecolor('#141414')
    metabric_axes2[1, 1].grid(alpha=0.2, axis='y')

if 'mutation_count' in metabric_clean.columns:
    mut_data = metabric_clean['mutation_count']
    metabric_axes2[1, 2].hist(mut_data, bins=30, color='#f72585', edgecolor='#ffffff', 
                              linewidth=0.5, alpha=0.8)
    metabric_axes2[1, 2].axvline(mut_data.median(), color='#00d9ff', linestyle='--', 
                                 linewidth=2, label=f'Median: {mut_data.median():.0f}')
    metabric_axes2[1, 2].set_xlabel('Mutation Count', color='#ffffff')
    metabric_axes2[1, 2].set_ylabel('Frequency', color='#ffffff')
    metabric_axes2[1, 2].set_title('Mutation Count Distribution', color='#ffffff', pad=10)
    metabric_axes2[1, 2].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes2[1, 2].set_facecolor('#141414')
    metabric_axes2[1, 2].grid(alpha=0.2)

plt.tight_layout()
plt.show()

print("✓ Feature engineering completed")

```

```python
print("\n" + "="*80)
print("SECTION 6: TARGET VARIABLES CREATION")
print("="*80 + "\n")

# Aggressiveness Score
if all(col in metabric_clean.columns for col in ['neoplasm_histologic_grade', 'tumor_stage', 
                                                   'lymph_nodes_examined_positive', 'nottingham_prognostic_index']):
    grade_norm = metabric_clean['neoplasm_histologic_grade'] / 3
    stage_norm = metabric_clean['tumor_stage'] / 4
    nodes_norm = metabric_clean['lymph_nodes_examined_positive'] / metabric_clean['lymph_nodes_examined_positive'].max()
    npi_norm = (metabric_clean['nottingham_prognostic_index'] - metabric_clean['nottingham_prognostic_index'].min()) / \
               (metabric_clean['nottingham_prognostic_index'].max() - metabric_clean['nottingham_prognostic_index'].min())
    
    metabric_clean['aggressiveness_score'] = (grade_norm * 3 + stage_norm * 3 + nodes_norm * 2 + npi_norm * 2)

# Growth Rate
if all(col in metabric_clean.columns for col in ['tumor_size', 'age_at_diagnosis', 'neoplasm_histologic_grade']):
    years_since_onset = np.maximum(metabric_clean['age_at_diagnosis'] - 40, 1)
    metabric_clean['growth_rate'] = (metabric_clean['tumor_size'] / years_since_onset) * \
                                     (metabric_clean['neoplasm_histologic_grade'] / 2)
    metabric_clean['growth_rate'] = metabric_clean['growth_rate'].clip(upper=50)

# Evolution 6M
if 'aggressiveness_score' in metabric_clean.columns:
    metabric_clean['evolution_6m'] = pd.cut(
        metabric_clean['aggressiveness_score'],
        bins=[-np.inf, 5, 7, np.inf],
        labels=[0, 1, 2]
    ).astype(int)

# Visualize target variables
metabric_fig3, metabric_axes3 = plt.subplots(1, 3, figsize=(18, 5))
metabric_fig3.patch.set_facecolor('#0a0a0a')
metabric_fig3.suptitle('Target Variables Distribution', fontsize=16, fontweight='bold', color='#ffffff')

if 'aggressiveness_score' in metabric_clean.columns:
    agg_data = metabric_clean['aggressiveness_score']
    metabric_axes3[0].hist(agg_data, bins=30, color='#ff1493', edgecolor='#ffffff', 
                           linewidth=0.5, alpha=0.8)
    metabric_axes3[0].axvline(agg_data.mean(), color='#00d9ff', linestyle='--', 
                              linewidth=2, label=f'Mean: {agg_data.mean():.2f}')
    metabric_axes3[0].set_xlabel('Aggressiveness Score', color='#ffffff')
    metabric_axes3[0].set_ylabel('Frequency', color='#ffffff')
    metabric_axes3[0].set_title('Aggressiveness Score (0-10)', color='#ffffff', pad=10)
    metabric_axes3[0].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes3[0].set_facecolor('#141414')
    metabric_axes3[0].grid(alpha=0.2)

if 'growth_rate' in metabric_clean.columns:
    growth_data = metabric_clean['growth_rate']
    metabric_axes3[1].hist(growth_data, bins=30, color='#9d4edd', edgecolor='#ffffff', 
                           linewidth=0.5, alpha=0.8)
    metabric_axes3[1].axvline(growth_data.median(), color='#00d9ff', linestyle='--', 
                              linewidth=2, label=f'Median: {growth_data.median():.2f}')
    metabric_axes3[1].set_xlabel('Growth Rate (mm/year)', color='#ffffff')
    metabric_axes3[1].set_ylabel('Frequency', color='#ffffff')
    metabric_axes3[1].set_title('Tumor Growth Rate', color='#ffffff', pad=10)
    metabric_axes3[1].legend(facecolor='#141414', edgecolor='#333333')
    metabric_axes3[1].set_facecolor('#141414')
    metabric_axes3[1].grid(alpha=0.2)

if 'evolution_6m' in metabric_clean.columns:
    evol_counts = metabric_clean['evolution_6m'].value_counts().sort_index()
    colors_evol = ['#00d9ff', '#ffa500', '#ff1493']
    bars = metabric_axes3[2].bar(['Stable (0)', 'Moderate (1)', 'Rapid (2)'], 
                                  [evol_counts.get(0, 0), evol_counts.get(1, 0), evol_counts.get(2, 0)],
                                  color=colors_evol, edgecolor='#ffffff', linewidth=0.5, alpha=0.8)
    metabric_axes3[2].set_ylabel('Count', color='#ffffff')
    metabric_axes3[2].set_title('Evolution at 6 Months', color='#ffffff', pad=10)
    metabric_axes3[2].set_facecolor('#141414')
    metabric_axes3[2].grid(alpha=0.2, axis='y')
    
    total = len(metabric_clean)
    for i, count in enumerate([evol_counts.get(0, 0), evol_counts.get(1, 0), evol_counts.get(2, 0)]):
        metabric_axes3[2].text(i, count, f'{count/total*100:.1f}%', ha='center', 
                               va='bottom', color='#ffffff', fontweight='bold')

plt.tight_layout()
plt.show()

print("✓ Target variables created")
```

    <h3 style="font-size: 20px; font-weight: 800; color: #ffffff; margin-bottom: 10px; border-left: 3px solid #ff1493; padding-left: 10px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif;">
        Hybrid Architecture: GRU-SVM
    </h3>
    <div style="background: #141414; padding: 15px; border-radius: 4px; border-left: 1px solid #333333;">
        <p style="font-size: 13px; color: #bbbbbb; line-height: 1.5; margin: 0;">
            <strong style="color: #ff1493;">Concept:</strong> Combining Deep Learning with Max-Margin Classification.
            <br><br>
            We reshape the 30 static features into a pseudo-sequence <code>(5 steps × 6 features)</code> to train a <strong>Gated Recurrent Unit (GRU)</strong>. Instead of using the GRU for the final prediction, we use it as a <strong>Feature Extractor</strong> to compress the data into 16 high-level latent features, which are then fed into an <strong>SVM</strong> for the final decision.
        </p>
    </div>

```python
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, GRU, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import warnings

warnings.filterwarnings('ignore')

print("\n" + "=" * 80)
print("🧠 HYBRID MODEL: GRU FEATURE EXTRACTOR + SVM CLASSIFIER")
print("=" * 80)

# ---------------------------------------------------------
# 1. DATA PREPARATION (MODEL-SPECIFIC)
# ---------------------------------------------------------
# We create specific copies for GRU to avoid breaking the original X_train used by other models
# Reshape logic: 30 features -> 5 timesteps x 6 features
X_train_gru = np.array(X_train).reshape(-1, 5, 6)
X_test_gru  = np.array(X_test).reshape(-1, 5, 6)

print(f"📊 Data Reshaped for GRU: {X_train_gru.shape}")

# ---------------------------------------------------------
# 2. FEATURE EXTRACTOR (GRU)
# ---------------------------------------------------------
# Define Input
gru_input = Input(shape=(5, 6), name='gru_input')

# GRU Layer to capture non-linear patterns
x = GRU(32, return_sequences=False, name='gru_layer')(gru_input)

# Dense Layer (The Latent Space we want to extract)
x = Dense(16, activation='relu', name='latent_features')(x)

# Output Head (Only used for training the feature extractor)
gru_output = Dense(1, activation='sigmoid', name='pretrain_output')(x)

# Compile & Train
feature_learner = Model(inputs=gru_input, outputs=gru_output, name='gru_feature_learner')
feature_learner.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])

print("🚀 Training GRU Feature Extractor...")
feature_learner.fit(X_train_gru, y_train, epochs=30, batch_size=32, verbose=0)

# ---------------------------------------------------------
# 3. FEATURE EXTRACTION
# ---------------------------------------------------------
# Create a truncated model that stops at the 'latent_features' layer
extractor = Model(inputs=feature_learner.input, outputs=feature_learner.get_layer('latent_features').output)

# Extract new high-level features
X_train_latent = extractor.predict(X_train_gru, verbose=0)
X_test_latent  = extractor.predict(X_test_gru, verbose=0)

print(f"✅ Features Extracted. New Shape: {X_train_latent.shape} (16 Latent Features)")

# ---------------------------------------------------------
# 4. FINAL CLASSIFICATION (SVM)
# ---------------------------------------------------------
# We feed the GRU-extracted features into a standard SVM
svm_hybrid = SVC(kernel='rbf', C=1.0, probability=True, random_state=42)
svm_hybrid.fit(X_train_latent, y_train)

# Predict
y_pred_hybrid = svm_hybrid.predict(X_test_latent)
y_prob_hybrid = svm_hybrid.predict_proba(X_test_latent)[:, 1]

# ---------------------------------------------------------
# 5. METRICS (SAVED FOR BENCHMARKING)
# ---------------------------------------------------------
acc_hybrid = accuracy_score(y_test, y_pred_hybrid)
sens_hybrid = recall_score(y_test, y_pred_hybrid)
tn, fp, fn, tp = confusion_matrix(y_test, y_pred_hybrid).ravel()

print("\n📊 GRU-SVM PERFORMANCE:")
print(f"   • Accuracy:    {acc_hybrid:.4f}")
print(f"   • Sensitivity: {sens_hybrid:.4f} (Recall)")
print(f"   • Missed Cases:{fn} (False Negatives)")
```

            <strong style="color: #00d9ff;">Objective:</strong> Establishing a robust probabilistic baseline. 
            <br><br>
            We implement <strong>Logistic Regression (Softmax)</strong> and perform a grid search over regularization strengths (<code>C</code>) and optimization solvers (<code>lbfgs</code>, <code>liblinear</code>, <code>newton-cg</code>) to find the configuration that yields the highest accuracy before moving to more complex architectures.
        </p>

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score
import warnings

warnings.filterwarnings('ignore')

print("\n" + "=" * 80)
print("📊 BASELINE MODEL: OPTIMIZED SOFTMAX REGRESSION")
print("=" * 80)

# ---------------------------------------------------------
# 0. SAFETY CHECK (Target Variable)
# ---------------------------------------------------------
# Ensure y is a 1D array to avoid shape errors
y_train_flat = y_train.ravel() if y_train.ndim > 1 else y_train
y_test_flat  = y_test.ravel()  if y_test.ndim > 1 else y_test

# ---------------------------------------------------------
# 1. MODEL SELECTION (Grid Search)
# ---------------------------------------------------------
configs_softmax = [
    {'C': 1.0, 'solver': 'lbfgs'}, {'C': 0.1, 'solver': 'lbfgs'},
    {'C': 10.0, 'solver': 'lbfgs'}, {'C': 1.0, 'solver': 'liblinear'},
    {'C': 1.0, 'solver': 'newton-cg'}
]

best_model_softmax = None
best_acc_softmax = 0

print("⚙️ Tuning Hyperparameters...")

for config in configs_softmax:
    # Train candidate
    model = LogisticRegression(**config, max_iter=3000, random_state=42)
    model.fit(X_train, y_train_flat)
    
    # Evaluate
    acc = model.score(X_test, y_test_flat)
    if acc > best_acc_softmax:
        best_acc_softmax = acc
        best_model_softmax = model

print(f"✅ Best Config Found: {best_model_softmax.get_params()['solver']} (C={best_model_softmax.get_params()['C']})")

# ---------------------------------------------------------
# 2. THRESHOLD OPTIMIZATION
# ---------------------------------------------------------
# We use distinct variable names (y_proba_softmax) to avoid conflicts
y_proba_softmax = best_model_softmax.predict_proba(X_test)[:, 1]

thresholds = np.arange(0.0, 1.01, 0.01)
best_threshold_softmax = 0.5
optimal_acc_softmax = 0

for t in thresholds:
    preds = (y_proba_softmax > t).astype(int)
    acc = accuracy_score(y_test_flat, preds)
    if acc > optimal_acc_softmax:
        optimal_acc_softmax = acc
        best_threshold_softmax = t

# ---------------------------------------------------------
# 3. FINAL EVALUATION (Saved to unique variables)
# ---------------------------------------------------------
# Apply optimal threshold
y_pred_softmax = (y_proba_softmax > best_threshold_softmax).astype(int)

# Calculate Metrics
tn_sm, fp_sm, fn_sm, tp_sm = confusion_matrix(y_test_flat, y_pred_softmax).ravel()
sens_softmax = tp_sm / (tp_sm + fn_sm)
spec_softmax = tn_sm / (tn_sm + fp_sm)

print("-" * 40)
print(f"🏆 SOFTMAX PERFORMANCE REPORT")
print("-" * 40)
print(f"Optimal Threshold : {best_threshold_softmax:.2f}")
print(f"Accuracy          : {optimal_acc_softmax:.4f}")
print(f"Sensitivity       : {sens_softmax:.4f} (Recall)")
print(f"Specificity       : {spec_softmax:.4f}")
print(f"Missed Cancers    : {fn_sm} (False Negatives)")
print("-" * 40)
```

            <strong style="color: #00d9ff;">Objective:</strong> Training a Linear Support Vector Machine using <strong>Stochastic Gradient Descent (SGD)</strong>. 
            <br><br>
            Unlike the standard SVM (which solves the dual problem quadratically), SGD allows for iterative learning, making it highly scalable and capable of online learning (updating weights as new patient data arrives).
        </p>

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import SGDClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, confusion_matrix

print("\n" + "=" * 80)
print("⚔️ CHALLENGER MODEL: SGD-SVM (LINEAR)")
print("=" * 80)

# =============================================================================
# 1. DATA PREPARATION (MODEL-SPECIFIC)
# =============================================================================
# [SAFETY CHECK] Ensure target is 1D
if y_train.ndim > 1 and y_train.shape[1] > 1:
    y_train_sgd = np.argmax(y_train, axis=1)
    y_test_sgd  = np.argmax(y_test, axis=1)
else:
    y_train_sgd = np.ravel(y_train)
    y_test_sgd  = np.ravel(y_test)

# [SCALING] SGD requires features to be scaled (Mean=0, Var=1)
# We use a local scaler to avoid modifying the global X_train
scaler_sgd = StandardScaler()
X_train_sgd = scaler_sgd.fit_transform(X_train)
X_test_sgd  = scaler_sgd.transform(X_test)

print(f"📊 Data Prepared for SGD: {X_train_sgd.shape}")

# =============================================================================
# 2. MODEL INITIALIZATION (SGD CLASSIFIER)
# =============================================================================
# Dynamic Alpha Calculation: 1 / (C * n_samples)
# Assuming a moderate regularization C=5
alpha_val = 1 / (5 * len(y_train_sgd))

sgd_model = SGDClassifier(
    loss='hinge',               # 'hinge' = Linear SVM
    penalty='l2',               # L2 Regularization
    alpha=alpha_val,            # Calculated learning rate/regularization
    class_weight='balanced',    # Handle class imbalance (Benign vs Malignant)
    learning_rate='constant',   # Stable convergence
    eta0=1e-3,                  # Initial learning rate
    max_iter=3000, 
    tol=1e-4,
    random_state=42,
    n_jobs=-1
)

# =============================================================================
# 3. TRAINING & PREDICTION
# =============================================================================
print(f"🚀 Training SGD-SVM (Alpha: {alpha_val:.6f})...")
sgd_model.fit(X_train_sgd, y_train_sgd)

# Predict
y_pred_sgd = sgd_model.predict(X_test_sgd)

# =============================================================================
# 4. EVALUATION
# =============================================================================
# Calculate metrics
acc_sgd  = accuracy_score(y_test_sgd, y_pred_sgd)
sens_sgd = recall_score(y_test_sgd, y_pred_sgd)
prec_sgd = precision_score(y_test_sgd, y_pred_sgd)
f1_sgd   = f1_score(y_test_sgd, y_pred_sgd)

# Confusion Matrix for False Negatives
tn_sgd, fp_sgd, fn_sgd, tp_sgd = confusion_matrix(y_test_sgd, y_pred_sgd).ravel()

print("\n📊 SGD-SVM PERFORMANCE:")
print("-" * 30)
print(f"   • Accuracy:       {acc_sgd:.4f}")
print(f"   • Sensitivity:    {sens_sgd:.4f} (Recall)")
print(f"   • Precision:      {prec_sgd:.4f}")
print(f"   • F1-Score:       {f1_sgd:.4f}")
print(f"   • Missed Cancers: {fn_sgd} (False Negatives)")
print("-" * 30)
```

```python
# =============================================================================
# ENHANCED TRIAGE: RISK + DANGER FLAGS (SIMPLE)
# =============================================================================

import pandas as pd
import numpy as np

# 1. Get original (unscaled) features
X_test_original = scaler_sgd.inverse_transform(X_test_scaled)

# 2. Create patient list
n_patients = len(y_proba_calibrated)
patient_ids = [f'P-{1000+i}' for i in range(n_patients)]

# 3. Define base risk stratification
def base_risk(prob):
    if prob >= 0.90: return '🔴 CRITICAL', 'Immediate biopsy'
    elif prob >= 0.80: return '🟠 HIGH', 'Imaging in 7 days'
    elif prob >= 0.60: return '🟡 MEDIUM', '6-month follow-up'
    elif prob >= 0.40: return '🟢 LOW', 'Routine screening'
    else: return '⚪ VERY LOW', 'Normal'

# 4. Count danger flags per patient
def count_flags(patient_idx):
    x = X_test_original[patient_idx]
    count = 0
    features = []
    for feat in feature_names:
        val = x[feature_names.index(feat)]
        thr = danger_thresholds[feat]
        if (thr['direction'] == 'HIGH' and val > thr['threshold']) or \
           (thr['direction'] == 'LOW' and val < thr['threshold']):
            count += 1
            features.append(feat)
    return count, features

# 5. Build enhanced patient list
records = []
for i in range(n_patients):
    prob = y_proba_calibrated[i]
    risk, action = base_risk(prob)
    flag_count, flagged = count_flags(i)
    
    # Upgrade risk if ≥2 danger flags
    if flag_count >= 2:
        if risk == '🟡 MEDIUM':
            risk, action = '🟠 HIGH', 'Imaging in 7 days [↑ 2+ danger features]'
        elif risk == '🟢 LOW':
            risk, action = '🟡 MEDIUM', '6-month follow-up [↑ 2+ danger features]'
        elif risk == '⚪ VERY LOW':
            risk, action = '🟢 LOW', 'Routine screening [↑ 2+ danger features]'
    
    records.append({
        'Patient_ID': patient_ids[i],
        'Probability': prob,
        'Risk_Group': risk,
        'Action': action,
        'Danger_Count': flag_count,
        'Danger_Features': flagged
    })

# 6. Create DataFrame and show priority list
triage_df = pd.DataFrame(records)

# Priority: High prob OR ≥2 danger flags
priority = triage_df[
    (triage_df['Probability'] >= 0.80) | (triage_df['Danger_Count'] >= 2)
].sort_values('Probability', ascending=False)

print("\n🏥 ENHANCED PRIORITY QUEUE")
print(priority[[
    'Patient_ID', 'Probability', 'Risk_Group', 'Action', 'Danger_Count'
]].head(10).to_string(index=False))
```

```python
print("\n" + "="*80)
print("SECTION 11: MODEL DEFINITION & TRAINING")
print("="*80 + "\n")

from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import r2_score, mean_absolute_error, accuracy_score

RANDOM_STATE = 42

# Define models to compare
models = {
    'Random Forest': MultiOutputRegressor(
        RandomForestRegressor(
            n_estimators=200,
            max_depth=10,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    ),
    'Gradient Boosting': MultiOutputRegressor(
        GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.1,
            max_depth=5,
            random_state=RANDOM_STATE
        )
    )
}

print("Models to train:")
for name in models.keys():
    print(f"  • {name}")

# Train and evaluate each model
results = {}

for name, model in models.items():
    print(f"\n{'='*60}")
    print(f"Training: {name}")
    print('='*60)
    
    # Training
    model.fit(X_train_scaled, y_train)
    print("✓ Model trained")
    
    # Predictions
    y_train_pred = model.predict(X_train_scaled)
    y_test_pred = model.predict(X_test_scaled)
    print("✓ Predictions generated")
    
    # Metrics for each target
    # Aggressiveness Score
    r2_aggr_train = r2_score(y_train.iloc[:, 0], y_train_pred[:, 0])
    r2_aggr_test = r2_score(y_test.iloc[:, 0], y_test_pred[:, 0])
    mae_aggr = mean_absolute_error(y_test.iloc[:, 0], y_test_pred[:, 0])
    
    # Growth Rate
    r2_growth_train = r2_score(y_train.iloc[:, 1], y_train_pred[:, 1])
    r2_growth_test = r2_score(y_test.iloc[:, 1], y_test_pred[:, 1])
    mae_growth = mean_absolute_error(y_test.iloc[:, 1], y_test_pred[:, 1])
    
    # Evolution 6M (classification)
    y_test_pred_class = np.round(y_test_pred[:, 2]).astype(int).clip(0, 2)
    acc_test = accuracy_score(y_test.iloc[:, 2], y_test_pred_class)
    
    # Display results
    print(f"\nAggressiveness Score:")
    print(f"  R² train: {r2_aggr_train:.4f}")
    print(f"  R² test:  {r2_aggr_test:.4f}")
    print(f"  MAE test: {mae_aggr:.4f}")
    
    print(f"\nGrowth Rate:")
    print(f"  R² train: {r2_growth_train:.4f}")
    print(f"  R² test:  {r2_growth_test:.4f}")
    print(f"  MAE test: {mae_growth:.4f}")
    
    print(f"\nEvolution 6M:")
    print(f"  Accuracy: {acc_test:.4f} ({acc_test*100:.1f}%)")
    
    # Store results
    results[name] = {
        'model': model,
        'r2_aggr_train': r2_aggr_train,
        'r2_aggr_test': r2_aggr_test,
        'mae_aggr': mae_aggr,
        'r2_growth_train': r2_growth_train,
        'r2_growth_test': r2_growth_test,
        'mae_growth': mae_growth,
        'acc_evol': acc_test,
        'predictions_train': y_train_pred,
        'predictions_test': y_test_pred
    }

print("\n✓ All models trained and evaluated")
```

```python
# ============================================================================
# SECTION 14: CONFUSION MATRIX ANALYSIS
# ============================================================================
print("\n" + "="*80)
print("SECTION 14: CONFUSION MATRIX ANALYSIS")
print("="*80 + "\n")

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Confusion matrix for Evolution 6M
y_pred_class = np.round(y_pred_final[:, 2]).astype(int).clip(0, 2)
cm = confusion_matrix(y_test.iloc[:, 2], y_pred_class)

print("Confusion Matrix - Evolution 6M:")
print(cm)

# Calculate per-class metrics
print("\nPer-Class Metrics:")
for i, class_name in enumerate(['Stable (0)', 'Moderate (1)', 'Rapid (2)']):
    true_positive = cm[i, i]
    false_positive = cm[:, i].sum() - true_positive
    false_negative = cm[i, :].sum() - true_positive
    true_negative = cm.sum() - (true_positive + false_positive + false_negative)
    
    precision = true_positive / (true_positive + false_positive) if (true_positive + false_positive) > 0 else 0
    recall = true_positive / (true_positive + false_negative) if (true_positive + false_negative) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    
    print(f"\n{class_name}:")
    print(f"  Precision: {precision:.3f}")
    print(f"  Recall:    {recall:.3f}")
    print(f"  F1-Score:  {f1:.3f}")

# Visualize confusion matrix
metabric_fig8, metabric_ax8 = plt.subplots(figsize=(10, 8))
metabric_fig8.patch.set_facecolor('#0a0a0a')

# Create custom colormap for dark theme
from matplotlib.colors import LinearSegmentedColormap
colors_list = ['#141414', '#ff1493']
n_bins = 100
cmap = LinearSegmentedColormap.from_list('custom', colors_list, N=n_bins)

im = metabric_ax8.imshow(cm, interpolation='nearest', cmap=cmap)
metabric_ax8.figure.colorbar(im, ax=metabric_ax8)

# Set ticks and labels
classes = ['Stable', 'Moderate', 'Rapid']
tick_marks = np.arange(len(classes))
metabric_ax8.set_xticks(tick_marks)
metabric_ax8.set_yticks(tick_marks)
metabric_ax8.set_xticklabels(classes, color='#ffffff')
metabric_ax8.set_yticklabels(classes, color='#ffffff')

# Add text annotations
thresh = cm.max() / 2.
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        metabric_ax8.text(j, i, format(cm[i, j], 'd'),
                         ha="center", va="center",
                         color="#ffffff" if cm[i, j] > thresh else "#ffffff",
                         fontsize=14, fontweight='bold')

metabric_ax8.set_ylabel('True Label', color='#ffffff', fontsize=12)
metabric_ax8.set_xlabel('Predicted Label', color='#ffffff', fontsize=12)
metabric_ax8.set_title(f'Confusion Matrix - Evolution 6M\n{best_model_name}', 
                       color='#ffffff', fontsize=14, fontweight='bold', pad=15)
metabric_ax8.set_facecolor('#141414')

plt.tight_layout()
plt.show()
```
