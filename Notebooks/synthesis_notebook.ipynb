```python
import pandas as pd

# 1. RELOAD DATA (Crucial to fix the current NaN error)
# We reload from source to ensure we start with 'M' and 'B' strings
try:
    df = pd.read_csv('../data/row/data.csv')
except FileNotFoundError:
    # Fallback if you are in a different directory structure
    df = pd.read_csv('data.csv') 

# 2. SAFE MAPPING
# We check if the column is text (object) before mapping. 
# If it's already numbers (from a previous run), we skip this step.
if df['diagnosis'].dtype == 'object':
    print("🔄 Mapping Diagnosis: 'M' -> 1, 'B' -> 0")
    df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
else:
    print("✅ Diagnosis is already numeric (Skipping mapping)")

# Drop noise columns
cols_to_drop = ['id', 'Unnamed: 32']
df = df.drop(columns=[c for c in cols_to_drop if c in df.columns], errors='ignore')

# 4. DEFINE X and y
X = df.drop('diagnosis', axis=1)
y = df['diagnosis']

print("-" * 30)
print("✅ Data Successfully Cleaned & Encoded")
print(f"Features (X): {X.shape}")
print(f"Target (y):   {y.shape}")
print("-" * 30)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Split: Hide 30% of data for testing (Simulating new, unseen patients)
# 'stratify=y' ensures the proportion of Cancer/Benign is consistent in both sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Scaling: Essential for SVM/SGD to treat all features equally (e.g. Area vs Smoothness)
scaler = StandardScaler()

# Fit ONLY on training data (to avoid cheating/leakage)
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("✅ Data Split & Scaled")
print(f"Training Set: {X_train_scaled.shape}")
print(f"Test Set:     {X_test_scaled.shape}")
```

```python
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, GRU, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import warnings

warnings.filterwarnings('ignore')

print("\n" + "=" * 80)
print("🧠 HYBRID MODEL: GRU FEATURE EXTRACTOR + SVM CLASSIFIER")
print("=" * 80)

# ---------------------------------------------------------
# 1. DATA PREPARATION (MODEL-SPECIFIC)
# ---------------------------------------------------------
# We create specific copies for GRU to avoid breaking the original X_train used by other models
# Reshape logic: 30 features -> 5 timesteps x 6 features
X_train_gru = np.array(X_train).reshape(-1, 5, 6)
X_test_gru  = np.array(X_test).reshape(-1, 5, 6)

print(f"📊 Data Reshaped for GRU: {X_train_gru.shape}")

# ---------------------------------------------------------
# 2. FEATURE EXTRACTOR (GRU)
# ---------------------------------------------------------
# Define Input
gru_input = Input(shape=(5, 6), name='gru_input')

# GRU Layer to capture non-linear patterns
x = GRU(32, return_sequences=False, name='gru_layer')(gru_input)

# Dense Layer (The Latent Space we want to extract)
x = Dense(16, activation='relu', name='latent_features')(x)

# Output Head (Only used for training the feature extractor)
gru_output = Dense(1, activation='sigmoid', name='pretrain_output')(x)

# Compile & Train
feature_learner = Model(inputs=gru_input, outputs=gru_output, name='gru_feature_learner')
feature_learner.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])

print("🚀 Training GRU Feature Extractor...")
feature_learner.fit(X_train_gru, y_train, epochs=30, batch_size=32, verbose=0)

# ---------------------------------------------------------
# 3. FEATURE EXTRACTION
# ---------------------------------------------------------
# Create a truncated model that stops at the 'latent_features' layer
extractor = Model(inputs=feature_learner.input, outputs=feature_learner.get_layer('latent_features').output)

# Extract new high-level features
X_train_latent = extractor.predict(X_train_gru, verbose=0)
X_test_latent  = extractor.predict(X_test_gru, verbose=0)

print(f"✅ Features Extracted. New Shape: {X_train_latent.shape} (16 Latent Features)")

# ---------------------------------------------------------
# 4. FINAL CLASSIFICATION (SVM)
# ---------------------------------------------------------
# We feed the GRU-extracted features into a standard SVM
svm_hybrid = SVC(kernel='rbf', C=1.0, probability=True, random_state=42)
svm_hybrid.fit(X_train_latent, y_train)

# Predict
y_pred_hybrid = svm_hybrid.predict(X_test_latent)
y_prob_hybrid = svm_hybrid.predict_proba(X_test_latent)[:, 1]

# ---------------------------------------------------------
# 5. METRICS (SAVED FOR BENCHMARKING)
# ---------------------------------------------------------
acc_hybrid = accuracy_score(y_test, y_pred_hybrid)
sens_hybrid = recall_score(y_test, y_pred_hybrid)
tn, fp, fn, tp = confusion_matrix(y_test, y_pred_hybrid).ravel()

print("\n📊 GRU-SVM PERFORMANCE:")
print(f"   • Accuracy:    {acc_hybrid:.4f}")
print(f"   • Sensitivity: {sens_hybrid:.4f} (Recall)")
print(f"   • Missed Cases:{fn} (False Negatives)")
```

            We implement <strong>Logistic Regression (Softmax)</strong> and perform a grid search over regularization strengths (<code>C</code>) and optimization solvers (<code>lbfgs</code>, <code>liblinear</code>, <code>newton-cg</code>) to find the configuration that yields the highest accuracy before moving to more complex architectures.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score
import warnings

warnings.filterwarnings('ignore')

print("\n" + "=" * 80)
print("📊 BASELINE MODEL: OPTIMIZED SOFTMAX REGRESSION")
print("=" * 80)

# ---------------------------------------------------------
# 0. SAFETY CHECK (Target Variable)
# ---------------------------------------------------------
# Ensure y is a 1D array to avoid shape errors
y_train_flat = y_train.ravel() if y_train.ndim > 1 else y_train
y_test_flat  = y_test.ravel()  if y_test.ndim > 1 else y_test

# ---------------------------------------------------------
# 1. MODEL SELECTION (Grid Search)
# ---------------------------------------------------------
configs_softmax = [
    {'C': 1.0, 'solver': 'lbfgs'}, {'C': 0.1, 'solver': 'lbfgs'},
    {'C': 10.0, 'solver': 'lbfgs'}, {'C': 1.0, 'solver': 'liblinear'},
    {'C': 1.0, 'solver': 'newton-cg'}
]

best_model_softmax = None
best_acc_softmax = 0

print("⚙️ Tuning Hyperparameters...")

for config in configs_softmax:
    # Train candidate
    model = LogisticRegression(**config, max_iter=3000, random_state=42)
    model.fit(X_train, y_train_flat)
    
    # Evaluate
    acc = model.score(X_test, y_test_flat)
    if acc > best_acc_softmax:
        best_acc_softmax = acc
        best_model_softmax = model

print(f"✅ Best Config Found: {best_model_softmax.get_params()['solver']} (C={best_model_softmax.get_params()['C']})")

# ---------------------------------------------------------
# 2. THRESHOLD OPTIMIZATION
# ---------------------------------------------------------
# We use distinct variable names (y_proba_softmax) to avoid conflicts
y_proba_softmax = best_model_softmax.predict_proba(X_test)[:, 1]

thresholds = np.arange(0.0, 1.01, 0.01)
best_threshold_softmax = 0.5
optimal_acc_softmax = 0

for t in thresholds:
    preds = (y_proba_softmax > t).astype(int)
    acc = accuracy_score(y_test_flat, preds)
    if acc > optimal_acc_softmax:
        optimal_acc_softmax = acc
        best_threshold_softmax = t

# ---------------------------------------------------------
# 3. FINAL EVALUATION (Saved to unique variables)
# ---------------------------------------------------------
# Apply optimal threshold
y_pred_softmax = (y_proba_softmax > best_threshold_softmax).astype(int)

# Calculate Metrics
tn_sm, fp_sm, fn_sm, tp_sm = confusion_matrix(y_test_flat, y_pred_softmax).ravel()
sens_softmax = tp_sm / (tp_sm + fn_sm)
spec_softmax = tn_sm / (tn_sm + fp_sm)

print("-" * 40)
print(f"🏆 SOFTMAX PERFORMANCE REPORT")
print("-" * 40)
print(f"Optimal Threshold : {best_threshold_softmax:.2f}")
print(f"Accuracy          : {optimal_acc_softmax:.4f}")
print(f"Sensitivity       : {sens_softmax:.4f} (Recall)")
print(f"Specificity       : {spec_softmax:.4f}")
print(f"Missed Cancers    : {fn_sm} (False Negatives)")
print("-" * 40)
```

            Unlike the standard SVM (which solves the dual problem quadratically), SGD allows for iterative learning, making it highly scalable and capable of online learning (updating weights as new patient data arrives).

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import SGDClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, confusion_matrix

print("\n" + "=" * 80)
print("⚔️ CHALLENGER MODEL: SGD-SVM (LINEAR)")
print("=" * 80)

# =============================================================================
# 1. DATA PREPARATION (MODEL-SPECIFIC)
# =============================================================================
# [SAFETY CHECK] Ensure target is 1D
if y_train.ndim > 1 and y_train.shape[1] > 1:
    y_train_sgd = np.argmax(y_train, axis=1)
    y_test_sgd  = np.argmax(y_test, axis=1)
else:
    y_train_sgd = np.ravel(y_train)
    y_test_sgd  = np.ravel(y_test)

# [SCALING] SGD requires features to be scaled (Mean=0, Var=1)
# We use a local scaler to avoid modifying the global X_train
scaler_sgd = StandardScaler()
X_train_sgd = scaler_sgd.fit_transform(X_train)
X_test_sgd  = scaler_sgd.transform(X_test)

print(f"📊 Data Prepared for SGD: {X_train_sgd.shape}")

# =============================================================================
# 2. MODEL INITIALIZATION (SGD CLASSIFIER)
# =============================================================================
# Dynamic Alpha Calculation: 1 / (C * n_samples)
# Assuming a moderate regularization C=5
alpha_val = 1 / (5 * len(y_train_sgd))

sgd_model = SGDClassifier(
    loss='hinge',               # 'hinge' = Linear SVM
    penalty='l2',               # L2 Regularization
    alpha=alpha_val,            # Calculated learning rate/regularization
    class_weight='balanced',    # Handle class imbalance (Benign vs Malignant)
    learning_rate='constant',   # Stable convergence
    eta0=1e-3,                  # Initial learning rate
    max_iter=3000, 
    tol=1e-4,
    random_state=42,
    n_jobs=-1
)

# =============================================================================
# 3. TRAINING & PREDICTION
# =============================================================================
print(f"🚀 Training SGD-SVM (Alpha: {alpha_val:.6f})...")
sgd_model.fit(X_train_sgd, y_train_sgd)

# Predict
y_pred_sgd = sgd_model.predict(X_test_sgd)

# =============================================================================
# 4. EVALUATION
# =============================================================================
# Calculate metrics
acc_sgd  = accuracy_score(y_test_sgd, y_pred_sgd)
sens_sgd = recall_score(y_test_sgd, y_pred_sgd)
prec_sgd = precision_score(y_test_sgd, y_pred_sgd)
f1_sgd   = f1_score(y_test_sgd, y_pred_sgd)

# Confusion Matrix for False Negatives
tn_sgd, fp_sgd, fn_sgd, tp_sgd = confusion_matrix(y_test_sgd, y_pred_sgd).ravel()

print("\n📊 SGD-SVM PERFORMANCE:")
print("-" * 30)
print(f"   • Accuracy:       {acc_sgd:.4f}")
print(f"   • Sensitivity:    {sens_sgd:.4f} (Recall)")
print(f"   • Precision:      {prec_sgd:.4f}")
print(f"   • F1-Score:       {f1_sgd:.4f}")
print(f"   • Missed Cancers: {fn_sgd} (False Negatives)")
print("-" * 30)
```

            A well-calibrated model ensures that a predicted <strong>70% risk</strong> actually corresponds to a <strong>70% true incidence rate</strong>. 
            If the model is overconfident, it might flag low-risk patients as high-priority (Resource Waste) or high-risk patients as low-priority (Danger).
        </p>
    </div>
